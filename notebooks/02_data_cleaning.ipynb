# Data Cleaning & Preparation

Objective:
Prepare the e-commerce dataset for analysis by handling missing values,
converting data types, removing duplicates, and creating derived features.

Date: 24-Sep-2026


```python
import pandas as pd

customers = pd.read_csv('../data/olist_customers_dataset.csv')
orders = pd.read_csv('../data/olist_orders_dataset.csv')
order_items = pd.read_csv('../data/olist_order_items_dataset.csv')
products = pd.read_csv('../data/olist_products_dataset.csv')

sales_data = (
    order_items
    .merge(products, on='product_id', how='left')
    .merge(orders, on='order_id', how='left')
    .merge(customers, on='customer_id', how='left')
)

print(sales_data.shape)
```

```python
sales_data.duplicated().sum()
```

```python
date_cols = [
    'shipping_limit_date',
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in date_cols:
    sales_data[col] = pd.to_datetime(sales_data[col])

sales_data[date_cols].dtypes
```

```python
sales_data['product_category_name'] = sales_data['product_category_name'].fillna('Unknown')

sales_data['product_name_lenght'] = sales_data['product_name_lenght'].fillna(
    sales_data['product_name_lenght'].median()
)

sales_data['product_description_lenght'] = sales_data['product_description_lenght'].fillna(
    sales_data['product_description_lenght'].median()
)

sales_data['product_photos_qty'] = sales_data['product_photos_qty'].fillna(
    sales_data['product_photos_qty'].median()
)
```

```python
sales_data['delivery_days'] = (
    sales_data['order_delivered_customer_date']
    - sales_data['order_purchase_timestamp']
).dt.days

sales_data['delivery_days'].describe()
```

```python
sales_data['order_year'] = sales_data['order_purchase_timestamp'].dt.year

sales_data['order_month'] = sales_data['order_purchase_timestamp'].dt.month

sales_data[['order_year', 'order_month']].head()
```

```python
sales_data['total_revenue'] = (
    sales_data['price']
    + sales_data['freight_value']
)

sales_data[['price', 'freight_value', 'total_revenue']].head()
```
